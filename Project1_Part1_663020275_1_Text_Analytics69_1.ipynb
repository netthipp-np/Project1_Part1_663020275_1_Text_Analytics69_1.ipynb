## เทคนิคที่อาจใช้ — เลือกเท่าที่จำเป็น

ไม่ต้องใช้ทั้งหมด โดยทั่วไป **2–4 เทคนิคก็เพียงพอ** ให้เลือกตามคำถามของทีม

| เทคนิค | เหมาะเมื่ออยากรู้ |
|---|---|
| **Word / n-gram frequency** | คนพูดถึงอะไรบ่อย |
| **TF-IDF / distinctive words** | สองกลุ่มพูดต่างกันอย่างไร |
| **Sentiment** | ข้อความเป็นบวก/ลบเพียงใด |

| **Topic modeling** | มีประเด็นอะไรที่เราอาจยังไม่รู้ |
| **Trend / segmentation** | ปัญหาเปลี่ยนตามเวลา หรือหนักในกลุ่มไหน |

> เทคนิคขั้นสูง เช่น embedding, clustering, regression, classification หรือ anomaly detection ทำได้ แต่ **ไม่บังคับ** และเหมาะสำหรับ Bonus เมื่อใช้ถูกและอธิบายได้

```python
# เตรียมข้อมูล: แปลงเวลา -> หาช่วงที่เก็บข้อมูลจริงของแต่ละไฟล์ -> ตัดโพสต์ที่เวลาผิดปกติ
df = raw.copy()
df["created_at"] = pd.to_datetime(df["created_at"], errors="coerce", utc=True)   # แปลงเป็นเวลา UTC

def collection_window(s):
    """ช่วงเวลาที่โพสต์หนาแน่นจริง = นาทีที่มีโพสต์ >= 10% ของนาทีที่ถี่ที่สุด"""
    per_min = s.dt.floor("min").value_counts()
    dense = per_min[per_min >= 0.1 * per_min.max()].index
    return pd.Series({"start": dense.min(), "end": dense.max() + pd.Timedelta(minutes=1)})

win = df.groupby("source_file")["created_at"].apply(collection_window).unstack()
df = df.join(win, on="source_file")
df["in_window"] = df["created_at"].between(df["start"], df["end"])   # True = อยู่ในช่วงเก็บข้อมูลจริง

posts = df[df["in_window"]].copy()                  # ตารางหลักที่ใช้วิเคราะห์ต่อ
posts["is_reply"] = posts["reply_to"].notna()       # โพสต์นี้เป็นการตอบกลับหรือไม่

summary = df.groupby("source_file").agg(
    ทั้งหมด=("uri", "size"),
    ใช้วิเคราะห์=("in_window", "sum"),
)
summary["ตัดทิ้ง"] = summary["ทั้งหมด"] - summary["ใช้วิเคราะห์"]
summary["ตัดทิ้ง_%"] = (summary["ตัดทิ้ง"] / summary["ทั้งหมด"] * 100).round(1)
summary["ช่วงเก็บจริง (UTC)"] = win["start"].dt.strftime("%H:%M") + "–" + win["end"].dt.strftime("%H:%M")
summary.loc["รวม"] = [summary["ทั้งหมด"].sum(), summary["ใช้วิเคราะห์"].sum(), summary["ตัดทิ้ง"].sum(),
                      round(summary["ตัดทิ้ง"].sum() / summary["ทั้งหมด"].sum() * 100, 1), ""]
summary.index = summary.index.str.replace("posts_", "").str.replace(".jsonl", "")
display(summary)

print(f"\nสรุป: ใช้วิเคราะห์ {len(posts):,} โพสต์ จาก {len(df):,} โพสต์ ({len(posts)/len(df):.1%})")
print(f"จำนวนบัญชีผู้ใช้ที่ไม่ซ้ำ: {posts['author'].nunique():,} บัญชี")
print(f"โพสต์ที่ตัดทิ้ง {(~df['in_window']).sum():,} โพสต์ คือโพสต์ที่ตั้งเวลาย้อนหลัง (ย้อนไปถึงปี {df['created_at'].min().year})")
```

## Q1 คนโพสต์เมื่อไร

```python
BKK = "Asia/Bangkok"

# คำนวณอัตราโพสต์ต่อนาที (Posts per minute) แยกตาม Collection Window
w = (posts.groupby("source_file")
          .agg(posts=("uri", "size"), start=("created_at", "min"), end=("created_at", "max"),
               reply_share=("is_reply", "mean"), image_share=("has_images", "mean")))
w["minutes"] = (w["end"] - w["start"]).dt.total_seconds() / 60
w["posts_per_min"] = w["posts"] / w["minutes"]
w["window_BKK"] = (w["start"].dt.tz_convert(BKK).dt.strftime("%H:%M") + "–"
                   + w["end"].dt.tz_convert(BKK).dt.strftime("%H:%M"))

print("=== สรุปอัตราการโพสต์แต่ละช่วงเวลา ===")
display(w[["window_BKK", "minutes", "posts", "posts_per_min", "reply_share", "image_share"]].round(3))
```

## Q2 ใช้ภาษาอะไร

```python
URL_RE  = re.compile(r"https?://\S+|www\.\S+")
MENT_RE = re.compile(r"@[\w.\-]+")
txt = posts["text"].fillna("").astype(str)
lang_text = txt.str.replace(URL_RE, "", regex=True).str.replace(MENT_RE, "", regex=True)

# 1. ตรวจภาษาตามระบบอักษร (Script Analysis)
SCRIPT = {
    "Thai": "[\u0E00-\u0E7F]",
    "Japanese": "[\u3040-\u30FF]",
    "Korean": "[\uAC00-\uD7AF]",
    "Chinese (Han)": "[\u4E00-\u9FFF]",
    "Arabic": "[\u0600-\u06FF]",
    "Cyrillic": "[\u0400-\u04FF]",
    "Latin": "[A-Za-z\u00C0-\u024F]",
}
cnt = pd.DataFrame({k: lang_text.str.count(p) for k, p in SCRIPT.items()})
dom = cnt.idxmax(axis=1)
dom[cnt.sum(axis=1) == 0] = "No text (emoji/link/number only)"
dom[cnt["Japanese"] > 0] = "Japanese"

# 2. แยกภาษาอักษรละตินด้วย Stopwords
LATIN = ["english", "spanish", "portuguese", "french", "german", "italian", "indonesian", "turkish", "dutch"]
sw_dict = {l: {w for w in stopwords.words(l) if len(w) >= 2} for l in LATIN}
owner = Counter(w for s in sw_dict.values() for w in s)
sw_dict = {l: {w for w in s if owner[w] == 1} for l, s in sw_dict.items()}

TOKEN_RE = re.compile(r"[a-z\u00e0-\u00f6\u00f8-\u00ff']+")
def detect_latin(t):
    toks = set(TOKEN_RE.findall(t.lower()))
    scores = {l: len(toks & s) for l, s in sw_dict.items()}
    best = max(scores.values())
    winners = [l for l, v in scores.items() if v == best]
    if best == 0 or len(winners) > 1: return "Latin-script (undetermined)"
    return winners[0].capitalize()

is_latin = dom == "Latin"
dom[is_latin] = lang_text[is_latin].map(detect_latin)
posts["lang"] = dom.values

# สรุปสัดส่วนภาษา
known = posts[~posts["lang"].isin(["No text (emoji/link/number only)", "Latin-script (undetermined)"])]
known_tab = (known["lang"].value_counts(normalize=True) * 100).round(2)
print("=== ภาษาหลัก 5 อันดับแรก ===")
display(known_tab.head(5).to_frame("share_%"))
```

## Q3 คนพูดถึงอะไร

```python
from IPython.display import HTML, display

# 1. แฮชแท็กเด่น
HASH_RE = re.compile(r"#\w*[^\W\d_]\w*")
posts["tags"] = txt.str.lower().str.findall(HASH_RE)
tag_df = posts[["uri", "author", "tags"]].explode("tags").dropna(subset=["tags"])
tag_tab = (tag_df.groupby("tags")
                 .agg(posts=("uri", "nunique"), authors=("author", "nunique"))
                 .sort_values("posts", ascending=False))

# 2. คำและวลีเด่นในภาษาอังกฤษ (Unigram & Bigram)
en = posts[posts["lang"] == "English"].copy()
en_txt = (en["text"].fillna("").astype(str)
            .str.replace(URL_RE, " ", regex=True)
            .str.replace(MENT_RE, " ", regex=True)
            .str.replace("#", " ", regex=True).str.lower())

STOP_EN = list(set(stopwords.words("english")) | {"amp", "www", "com", "bsky", "app", "https", "http", "just", "like", "get", "got"})

cv_uni = CountVectorizer(stop_words=STOP_EN, token_pattern=r"(?u)\b[a-z][a-z']+\b", ngram_range=(1,1), binary=True, min_df=30)
X_uni = cv_uni.fit_transform(en_txt)
top_unigrams = pd.Series(np.asarray(X_uni.sum(axis=0)).ravel(), index=cv_uni.get_feature_names_out()).sort_values(ascending=False).head(10)

cv_bi = CountVectorizer(stop_words=STOP_EN, token_pattern=r"(?u)\b[a-z][a-z']+\b", ngram_range=(2,2), binary=True, min_df=30)
X_bi = cv_bi.fit_transform(en_txt)
top_bigrams = pd.Series(np.asarray(X_bi.sum(axis=0)).ravel(), index=cv_bi.get_feature_names_out()).sort_values(ascending=False).head(10)

# ============================================================
# แสดงตารางเรียงกันเป็นแนวนอนด้วย HTML / Flexbox
# ============================================================
df_tags = tag_tab.head(5)
df_uni = top_unigrams.to_frame("posts")
df_bi = top_bigrams.to_frame("posts")

html_layout = f"""
<div style="display: flex; gap: 30px; align-items: flex-start;">
    <div>
        <h4 style="margin-bottom: 8px;">=== แฮชแท็กเด่น 5 อันดับแรก ===</h4>
        {df_tags.to_html()}
    </div>
    <div>
        <h4 style="margin-bottom: 8px;">=== คำเด่น (Unigrams) ===</h4>
        {df_uni.to_html()}
    </div>
    <div>
        <h4 style="margin-bottom: 8px;">=== วลีเด่น (Bigrams) ===</h4>
        {df_bi.to_html()}
    </div>
</div>
"""

display(HTML(html_layout))
```
